# urls_by_domain

Rolls Scrapy `stage1_discovery` URLs up by domain. That is the same shape as Data-visualizer `urls_by_domain` (`domain`, `url_count`, `file_types`, `last_crawl`, `success_rate`).

- Lake: `scrapy:Scraping_project/data/delta_lake/stage1_discovery`
- Visualizer: `data-visualizer:/api/datasets/urls_by_domain`
- Writes: `datasets/samples/urls_by_domain.csv`


```python
import csv
from pathlib import Path

OUT = Path("datasets/samples/urls_by_domain.csv")
FIELDS = ["domain", "url_count", "file_types", "last_crawl", "success_rate"]
# Sample rows standing in for a group-by over stage1_discovery.url.
ROWS = [
    {"domain": "example.com", "url_count": 128, "file_types": 3, "last_crawl": "2026-01-15T12:00:00Z", "success_rate": 96.1},
    {"domain": "news.example.org", "url_count": 54, "file_types": 2, "last_crawl": "2026-02-02T08:30:00Z", "success_rate": 88.9},
    {"domain": "data.example.net", "url_count": 17, "file_types": 4, "last_crawl": "2026-03-11T16:45:00Z", "success_rate": 100.0},
]

OUT.parent.mkdir(parents=True, exist_ok=True)
with OUT.open("w", newline="") as handle:
    writer = csv.DictWriter(handle, fieldnames=FIELDS)
    writer.writeheader()
    writer.writerows(ROWS)
print(f"wrote {len(ROWS)} rows to {OUT}")
```
