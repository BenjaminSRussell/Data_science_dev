# page_metadata

Projects Scrapy `stage2_page_analysis` (title, word count, quality, processed time) into the shared page-metadata sample. Data-visualizer exposes the joined form as `page_metadata_with_urls`.

- Lake: `scrapy:Scraping_project/data/delta_lake/stage2_page_analysis`
- Visualizer: `data-visualizer:/api/datasets/page_metadata_with_urls`
- Writes: `datasets/samples/page_metadata.csv`


```python
import csv
from pathlib import Path

OUT = Path("datasets/samples/page_metadata.csv")
FIELDS = ["url", "domain", "title", "word_count", "quality_score", "processed_at"]
ROWS = [
    {"url": "https://example.com/a", "domain": "example.com", "title": "Example A", "word_count": 420, "quality_score": 0.82, "processed_at": "2026-01-15T12:05:00Z"},
    {"url": "https://news.example.org/b", "domain": "news.example.org", "title": "News B", "word_count": 1280, "quality_score": 0.67, "processed_at": "2026-02-02T08:40:00Z"},
    {"url": "https://data.example.net/c", "domain": "data.example.net", "title": "Data C", "word_count": 90, "quality_score": 0.91, "processed_at": "2026-03-11T16:50:00Z"},
]

OUT.parent.mkdir(parents=True, exist_ok=True)
with OUT.open("w", newline="") as handle:
    writer = csv.DictWriter(handle, fieldnames=FIELDS)
    writer.writeheader()
    writer.writerows(ROWS)
print(f"wrote {len(ROWS)} rows to {OUT}")
```
