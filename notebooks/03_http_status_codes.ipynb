# http_status_codes

Counts `http_status_code` on the Scrapy `error_log` lake table. Data-visualizer serves the same histogram as `status_codes_distribution`.

- Lake: `scrapy:Scraping_project/data/delta_lake/error_log`
- Visualizer: `data-visualizer:/api/datasets/status_codes_distribution`
- Writes: `datasets/samples/http_status_codes.csv`


```python
import csv
from pathlib import Path

OUT = Path("datasets/samples/http_status_codes.csv")
FIELDS = ["status_code", "count", "percentage"]
ROWS = [
    {"status_code": 200, "count": 900, "percentage": 90.0},
    {"status_code": 404, "count": 70, "percentage": 7.0},
    {"status_code": 500, "count": 30, "percentage": 3.0},
]

OUT.parent.mkdir(parents=True, exist_ok=True)
with OUT.open("w", newline="") as handle:
    writer = csv.DictWriter(handle, fieldnames=FIELDS)
    writer.writeheader()
    writer.writerows(ROWS)
print(f"wrote {len(ROWS)} rows to {OUT}")
```
